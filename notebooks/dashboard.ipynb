```python
%load_ext autoreload
%autoreload 2
import sys
sys.path.append('../src')
    
from pathlib import Path
import polars as pl
from matplotlib import pyplot as plt
import numpy as np

import holoviews as hv
import panel as pn
hv.extension('bokeh')
#pn.extension()
pn.extension('tabulator')

plato_data_directory = Path('/STER/mamsie/tess/TIC/plato_fov/')
```

```python
df_hv = pl.read_parquet('viz.parquet').with_columns(
    pl.col('GaiaID').cast(pl.String),
    (10**pl.col('ls_frequency_fundamental')).alias('ls_frequency_fundamental'),
    pl.col('path').str.split('-').list.get(3).alias('sector')
).to_pandas()
df_hv['class'] = df_hv['class'].astype('category')
```

```python
from frequency_domain_features import compute_lombscargle_periodogram
from time_domain_features import acf_fft_from_time_flux
from light_curves import read_fits_lc, preprocess_lc

points = hv.Points(
    df_hv, kdims=['lambda_1', 'lambda_2'], vdims=['class']
).opts(tools=['tap'], color='class', width=1000, height=800, size=3, alpha= 0.75, cmap='tab10', legend_position='top')
pointer = hv.streams.Tap(source=points)
x = df_hv['lambda_1'].to_numpy()
y = df_hv['lambda_2'].to_numpy()

def nearest_index(xpos, ypos):
    if xpos is None or ypos is None:
        return 0  # empty until mouse enters plot
    d2 = (x - xpos)**2 + (y - ypos)**2
    return int(np.argmin(d2))

tab = pn.widgets.Tabulator(
    df_hv.drop(columns=["path"]),   # initial
    selectable=False,
    pagination=None,
    height=60,
)

def update_from_tap(event=None):
    idx = nearest_index(pointer.x, pointer.y)
    tab.value = df_hv.iloc[[idx]].drop(columns=["path"])
    
pointer.param.watch(lambda e: update_from_tap(), "x")
pointer.param.watch(lambda e: update_from_tap(), "y")

def plot_lc(time, flux, width, height):
    if time is None:
        return hv.Scatter(([], []), 'Time', 'Flux').opts(framewise=True, axiswise=True, width=width, height=height) 
    return hv.Scatter((time, flux), 'Time', 'Flux').opts(framewise=True, axiswise=True, width=width, height=height)

def plot_folded(time, flux, freq, width, height):
    if time is None:
        return hv.Scatter(([], []), 'Phase', 'Flux').opts(framewise=True, axiswise=True, width=width, height=height)
    phase = np.mod(time, 2/freq)*freq
    return hv.Scatter((phase, flux), 'Phase', 'Flux').opts(framewise=True, axiswise=True, width=width, height=height)

def plot_ls(time, flux, width, height):
    if time is None:
        return hv.Curve(([], []), 'Frequency', 'Amplitude').opts(framewise=True, axiswise=True, width=width, height=height, tools=['hover'])
    freq, ampl, _ = compute_lombscargle_periodogram(time, flux, oversampling_factor=1)
    return hv.Curve((freq, ampl), 'Frequency', 'Amplitude').opts(framewise=True, axiswise=True, width=width, height=height, tools=['hover'])

def plot_acf(time, flux, width, height):
    if time is None:
        return hv.Curve(([], []), 'Lag', 'ACF').opts(framewise=True, axiswise=True, width=width, height=height, tools=['hover'])
    lag, acf, _ = acf_fft_from_time_flux(time, flux, max_lag_days=4.0)
    return hv.Curve((lag, acf), 'Lag', 'ACF').opts(framewise=True, axiswise=True, width=width, height=height, tools=['hover'])


def plot_data(index, flux_column='cal_aper_flux', preprocessed=True, width=600, height=200):
    try:
        row = df_hv.iloc[[index]]
        time, flux, sector = read_fits_lc(row['path'].item(), flux_column=flux_column)
        if preprocessed == True:
            time, flux = preprocess_lc(time, flux, sector)    
        freq = row['ls_frequency_fundamental'].item()
    except:
        time, flux, freq = None, None, None        
    return hv.Layout([
        plot_lc(time, flux, width=width, height=height), 
        plot_folded(time, flux, freq, width=width, height=height), 
        plot_ls(time, flux, width=width, height=height), 
        plot_acf(time, flux, width=width, height=height)
    ]).cols(1)

def plot_data_from_pos(x, y):
    return plot_data(nearest_index(x, y))

dmap_lc = hv.DynamicMap(plot_data_from_pos, streams=[pointer])
layout = pn.Row(pn.Column(tab, points), dmap_lc)
layout
#points
```
